### PyTorch Tensor Data Types

| Data Type | Dtype | Description |
| :--- | :--- | :--- |
| **32-bit Floating Point** | `torch.float32` / `torch.float` | Standard floating-point type for most deep learning tasks. Balances precision and memory. |
| **64-bit Floating Point** | `torch.float64` / `torch.double` | Double-precision floating point for high-precision tasks. |
| **16-bit Floating Point** | `torch.float16` / `torch.half` | Half-precision for mixed-precision training to save memory on GPUs. |
| **BFloat16** | `torch.bfloat16` | Brain floating-point format used in mixed-precision training (TPUs/GPUs). |
| **8-bit Floating Point** | `torch.float8` | Ultra-low-precision for extreme memory-constrained setups. |
| **8-bit Integer** | `torch.int8` | Signed 8-bit integer used for model quantization during inference. |
| **16-bit Integer** | `torch.int16` / `torch.short` | Signed 16-bit integer for intermediate precision tasks. |
| **32-bit Integer** | `torch.int32` / `torch.int` | Standard signed integer type used for indexing and general calculations. |
| **64-bit Integer** | `torch.int64` / `torch.long` | Long integer, default type for target labels and large indices. |
| **8-bit Unsigned Integer** | `torch.uint8` | Unsigned 8-bit integer, commonly used for image pixel values (0–255). |
| **Boolean** | `torch.bool` | Stores `True` or `False`, used for logical masking operations. |
| **Complex 64** | `torch.complex64` | Complex numbers with 32-bit real & 32-bit imaginary parts. |
| **Complex 128** | `torch.complex128` | High-precision complex numbers with 64-bit real & 64-bit imaginary parts. |
| **Quantized Integer** | `torch.qint8` | Quantized signed 8-bit integer for inference. |
| **Quantized Unsigned Integer** | `torch.quint8` | Quantized unsigned 8-bit integer for image tasks. |

## **Inplace Operation**

```python
m=torch.rand(2,3)
n=torch.rand(2,3)
print(m)
print(n)
```

```python
m+n # completely a new tensor
```

```python
m.add_(n) # it means the operation modifies m itself (in-place).
```

```python
m
```

```python
torch.relu(m)  # it is completely a new tensor
```

```python
m.relu_()  # it means the operation modifies m itself (in-place).
```

```python
#In-place operation means an operation that directly modifies the original tensor
#instead of creating a new tensor.
#relu() → creates a new tensor
#relu_() → modifies the original tensor in-place
#put after any function_() -> modifies the original tensor in-place
```

```python
device=torch.device('cuda')
```

```python
#moving an existing tensor to GPU
a=torch.rand(2,3)
a
```

```python
b=a.to(device)  #this tensor made on GPU
```

```python
b+5
```

```python
import time
#define the size of the  matrices
size=10000 # large size for performance comparison

# create random matrices on CPU
matrix_cpu1=torch.randn(size,size)
matrix_cpu2=torch.randn(size,size)

#measure the CPU
start_time=time.time()
result_cpu=torch.matmul(matrix_cpu1,matrix_cpu2)  # matrix multiplication on CPU
cpu_time=time.time() - start_time
print(f"Time on CPU: {cpu_time:.4f} seconds")

#move matrices to GPU
matrix_gpu1=matrix_cpu1.to('cuda')
matrix_gpu2=matrix_cpu2.to('cuda')

#measure the time
start_time=time.time()
result_gpu=torch.matmul(matrix_gpu1,matrix_gpu2)   # matrix multiplication on GPU
torch.cuda.synchronize() # ensures all GPU operations are completely
gpu_time=time.time() - start_time
print(f"Time on GPU: {gpu_time:.4f} seconds")

#compare results
print("\nSpeedup (CPU time / GPU time):", cpu_time / gpu_time)
```
